```python
def createHistogram3(columnName, graphTitle, filterData=None):

    tempDF = df

    if filterData:
        tempDF = df[df[columnName] < 100]

    fig, ax = plt.subplots(figsize=(9, 4))

    median = tempDF[columnName].median()
    mean = tempDF[columnName].mean()

    sns.histplot(data=tempDF, x=columnName, bins=20, ax=ax)

    ax.axvline(median, color="deeppink", linestyle="--", label="Median")
    ax.axvline(mean, color="royalblue", label="Mean")

    ax.legend(frameon=False)
    ax.set_xlabel(columnName)
    ax.set_ylabel("Count")
    ax.set_title(graphTitle)

    plt.show()
```
